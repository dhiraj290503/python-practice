# Arguments and parameters

```python
# parameters
# values provided to the function while creating function
# the thing you accept is parameters
# the thing you provide to parameters
def sum(a,b):
    print(f"Sum of {a} and {b} is : {a+b}")

sum(10,19)
sum(10,200)
sum(25,20)
```

```python
def sum(name,age):
    print(f"name: {name} age:{age}")

sum(age=10,name="DJ")
sum(10,"DJ")
```

```python
def sum(a,b=20):  #default parameter
    print(f"Sum of {a} and {b} is : {a+b}")

sum(10,19) # default argument replaced (19)
```

# palindrome

```python
def pallindrome(st):
    st=st.upper()
    rev=''
    for i in range(len(st)-1,-1,-1):
        rev=rev+st[i]

    if rev == st:
        print(f"{st} is Pallindrome")
    else :
        print(f"{st} is  Not Pallindrome")

pallindrome("NAMAN")        
pallindrome("Nitin")        
pallindrome("NitiN")        
```
