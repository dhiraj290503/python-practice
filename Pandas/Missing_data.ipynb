# Missing Data 

```python
import numpy as np 
import pandas as pd 
```

## Finding Missing Data

```python
data={
    'A':[1,2,np.nan,4,5],
    "B":[np.nan,2,3,4,5],
    "C":[np.nan,2,3,4,5],
    "D":[1,np.nan,np.nan,np.nan,5],
    
}
```

```python
df=pd.DataFrame(data)
df
```

```python
df.isna()
```

```python
df.isna().sum()
```

```python
df.isna().any()
```

## removing missing Data 

```python
df
```

```python
df.dropna()
```

```python
df
```

```python
df.dropna(thresh=3)
```

# filling the missing data 

```python
df
```

```python
df.fillna(0)
```

```python
values ={"A":0,"B":100,"C":1000,"D":10000 }
df.fillna(value=values)
```
