# merging 2 dataframes

```python
import numpy as np 
import pandas as pd 
```

```python
emp={
    "emp_id":[1,2,3,4,5],
    "name":["dhiraj","aman","suyash","sohan","pratik"],
    "dep":["HR","IT","Finance","IT","HR"]
}

salaries={
    "emp_id":[1,2,3,6,7],
    "salary":[60000,50000,55000,70000,90000],
    "bonus":[500,600,700,800,900]
    
    
}

emp=pd.DataFrame(emp)
salaries=pd.DataFrame(salaries)
```

```python
emp
```

```python
salaries
```

```python
pd.merge(emp,salaries)
```

```python
pd.merge(emp,salaries,on="emp_id")
```

```python
pd.merge(emp,salaries,on="emp_id",how="inner")
```

```python
pd.merge(emp,salaries,on="emp_id",how="outer")
```

```python
emp
```

```python
salaries
```

```python
pd.merge(emp,salaries,on="emp_id",how="left")
```

```python
pd.merge(emp,salaries,on="emp_id",how="right")
```

# concatinaition

```python
df1=pd.DataFrame({
    "A":["A0","A1","A2"],
    "B":["B0","B1","B2"],
    "C":["C0","C1","C2"]
}) 

df2=pd.DataFrame({
    "A":["A3","A4","A5"],
    "B":["B3","B4","B5"],
    "C":["C3","C4","C5"],
})
```

```python
df1
```

```python
df2
```

```python
pd.concat([df1,df2])
```

```python
pd.concat([df2,df1])
```

```python
pd.concat([df1,df2],axis=1)
```

**joining 2 data frames** 

```python
df1=pd.DataFrame({
    "name":["dhiraj","shekhar","sam"]
},index=[1,2,3])

df2=pd.DataFrame({
    "score":[85,90,70]
},index=[2,3,4])
```

```python
df1
```

```python
df2
```

```python
df1.join(df2)
```

```python
df1.join(df2,how="outer")
```

```python
df1.join(df2,how="inner")

```
