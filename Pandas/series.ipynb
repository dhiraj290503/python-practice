# Panda Series

```python
A one dimensional labeled array capable of holding and data types.
the axis lables are collectively called index
```

## creating series in many ways 

```python
import numpy as np
import pandas as pd
```

```python
labels = ["a","b","c"]
list1=[10,20,30]

arr=np.array([10,20,30])
d = {1:10,2:20,3:30}
```

```python
pd.Series(list1)
```

```python
pd.Series(list1,index=labels)
```

```python
pd.Series(arr)
```

```python
pd.Series(arr,index=labels)
```

```python
pd.Series(d)
```
