```python

import pandas as pd

data = {
    "category": ["Electronics", "Clothing", "Grocery", "Electronics",
                 "Furniture", "Clothing", "Grocery", "Furniture"],

    "store": ["Pune", "Mumbai", "Pune", "Nashik",
              "Mumbai", "Pune", "Nashik", "Pune"],

    "sales": [25000, 12000, 8000, 30000,
              45000, 15000, 9500, 38000],

    "quantity": [5, 12, 30, 6, 3, 15, 25, 4],

    "date": pd.date_range("2026-01-01", periods=8)
}

df = pd.DataFrame(data)

df
```

```python
# group by category and sum of sales

cat=df.groupby("category")["sales"].sum()
cat
```

```python
# group by store and calculate sum 
store=df.groupby("store")["sales"].sum()
store
```

```python
group by multiple columns
group by categories and store
```

```python
cat=df.groupby(["category","store"])["sales"].sum()
cat
```
