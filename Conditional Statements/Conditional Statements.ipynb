# Conditional Statements

```python
>It allows python for decision making executing different block of code based on conditions
    task a < 10
    task b > 10
> if --> executes only when the if condition is true
> if-else --> executes when the if condition is true ortherwise executes else 
> elif --> to write multiple if conditions to execute code if is executed if true or else moved to next block of code where we have elif ,if 
        the elif condition is true then it will execute or check for another elif condition or else then it will execute  else part 
```

```python
a = 12 
b = 15

if a > 10:
    print(f"{a} is greater than 10")
```

```python
# even or odd

num=int(input("Enter a number"))

if num%2==0:
    print(f"{num} is even")
else :
    print(f"{num} is odd")
```

```python
year = int(input("Enter Year:"))

if (year % 4 == 0 and year % 100 != 0) or (year % 400 == 0)    :
    print(f"{year} is leap year")
else :
    print("not a leap year")
                 
```

```python
# take the iunput  of temperature in celsius and check conditions 

temp= int(input("enter temperature in Celsius"))

if temp < 0 :
    print("Below 0°C → Freezing Cold")
elif temp >0 and temp <= 10:
    print("0°C to 10°C → very cold")
elif temp > 10 and temp <=20 :
    print("10°C to 20°C → cold")
elif temp >20 and temp <=30:
    print("20°C to 30°C → Pleasant")
elif temp > 30 and temp <=40:
    print("30°C to 40°C → Hot b")
elif temp > 40 :
    print("Above 40°C → Very Hot ")
                
```
