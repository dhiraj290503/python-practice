# Sets Methods

## remove

```python
s={1,2,3,4,5,6,7}
s.remove(2)
print(2)
print(s)
```

## pop

```python
s.pop()
print(s)
```

```python
a={1,2,3}
b={3,4,5}

print(a)
print(b)
```

## union

```python
s=a.union(b)
print(s)
print(a|b)
```

## intersection

```python
s=a.intersection(b)
print(s)
print(a&b)
```

```python
s=a.difference(b)
print(s)
print(a-b)
```

## symetric difference 

```python
s=a.symmetric_difference(b)
print(s)
print(a^b)
```
