# Arithmetic

```python
It performs mathematical operations like addition division multiplication substraction etc.
    
```

```python
a=12
b=6

addition = a+b #aditiion
sub= a-b #substraction
multi= a*b #multiplication
div= a/b #division 
fd = a//b #floor division
mod = a%b #modulus
power = a**b #exponentiation/power

print(f"Addition is {addition}")
print(f"Substration is {sub}")
print(f"Multiplication is {multi}")
print(f"Division is {div}")
print(f"Floor division is {fd}")
print(f"Modulus  is {mod}")
print(f"Power of {a} raise to {b} is {power}")
```

# Comparision Operators

# Logical Operators

```python
logical operators in python  are used to combine multiple conditions and return boolean results (True or False)

three types
    1.and >> return true if both conditions are ture
    2.or  >> return true if any one condition is true 
    3.not >> reverse the boolean value or result
```

```python
print(123>20 and 34==43 and 40<20)
```

```python
print(123>20 and 34<=43 and 20==20)
```

```python
# or
print(12==12 or 123>231)
```

```python
print(20>30 or 12==12)
```

```python
print(not 12==12)
```

```python
print((1==1)!=(1>2))
```

```python
print((1==1) == (10>2))
```
