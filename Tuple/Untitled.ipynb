# Tuples


```python
>tuples are mutable
>duplicate values are present
>ordered and can access them using index
>heterogenous
```

```python
tuple1=(1,2,3,4,5,"dhiraj",True,"hii",3.14)
print(type(tuple1))
tuple1
```

```python
tuple1[0]=22
```

```python
for i in tuple1:
    print(i)
```

```python
for i in range(len(tuple1)):
    print(tuple1[i])
```

## Methods

```python
# index()

print(tuple1.index(4))
```

```python
# count()

print(tuple1.count(2))
```
