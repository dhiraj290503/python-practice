```python
arr[1]
```

```python
arr[2]
```

```python
arr[3]
```

```python
arr[0][2]
```

```python
arr[2][2]
```

## slicing

```python

arr[0:3,1:3] #[row,#col]
```

```python
arr=np.arange(1,31).reshape(6,5)
arr
```

```python
arr[3:6,3:5]
```

# Boolean Indexing 

```python
arr=np.arange(11,21)
arr
```

```python
# even numbers
bool_index=arr % 2 ==0
bool_index
```

```python
arr = arr[bool_index]
arr
```

```python
arr=np.arange(11,21)
arr
odd_ele= arr % 2 != 0
odd_ele
```

```python
arr=arr[odd_ele]
arr
```
