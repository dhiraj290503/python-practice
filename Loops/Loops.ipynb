# Break and continue

```python
break to stop the loop
continue is skipping the current state and continuing the loop
```

```python
for i in range (1,11):
    if i==5:
        print("break executed")
        break;
    else:
        print(i)
    
```

```python
for i in range (1,11):
    if i==5:
        continue;
    print(i)
    
```

```python

# sum of all even and odd numbers in range

n=int(input("enter a number"))
even=0
odd=0

for i in range(1,n+1):
    if i%2==0:
        even+=i
    else:
        odd+=i
print(f"Sum of even numbers from 1 to {n}: {even}")
print(f"Sum of odd numbers from 1 to {n}: {odd}")
```

```python
# Print all the factors of a Number

n=int(input("Enter a number"))

for i in range(1,n+1):
    if n % i == 0:
        print(f"factorial of {n} are {i}")
```

```python
# perfect number
```

```python
n=int(input("Enter a number:"))
sum =0

for i in range(1,n):
    if n % i ==0:
        sum+=i
        # print(f"sum of {n} factorial is : {sum}")
if sum == n:
    print(f"{n} is perfect number")
else :
    print(f"{n} is not a perfect Number")
    
```

```python
# Prime Number

n=int(input("enter number"))
count=0

for i in range(1,n+1):
    if n % i==0:
        count+=1

if count == 2:
    print(f"{n} is prime number")
else :
    print(f"{n} is Not prime number")
```

```python
# Pallindrome

a=input("Enter yopur name ").upper()
b=""
for i in range(len(a)-1,-1,-1):
    b+=a[i]

if b==a:
    print(f"{a} is Pallindrome")
else:
    print(f"{a} is Not Pallindrome")
```

```python
# Count all letters,digits and special symbols from given string
# string = P@#yn26at^&i5ve

str1="P@#yn26at^&i5ve"
char=0
dig=0
spchar=0

for i in str1:
    if i.isdigit():
        dig+=1
    elif i.isalpha():
        char+=1
    else:
        spchar+=1

print(f"Your digits are {dig}\nalphabetes are {char}\nspecial characters are {spchar}")
```
