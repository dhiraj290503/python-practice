```python
import random

num = random.randint(1, 10)
tries=0

while True:
    
    num2=int(input("Predict a Number between 1 to 10:"))

    if num == num2:
        tries+=1
        print(f"{num} is same what you Predicted in {tries} tries")
        break
    elif num < num2:
        print("go a little Lower")
        tries+=1
    elif num > num2:
        print("go a little higher")   
        tries+=1 
    else :
        tries+=1
        print("You are Wrong")
    
```
