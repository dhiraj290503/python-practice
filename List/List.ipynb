```python
# mean of list
list1=[1,2,3,2,56,77,46,85]
length=len(list1)
sum=0
for i in list1:
    sum+=i

print(f"total is {sum}")
print(f"mean is {sum/length}")
```

```python
# greatest element and print its index

list1=[12,45,64,12,556,78,65,32]
max=list1[0]
index=0

for i in range(len(list1)):
    
    if list1[i] > max:
        max=list1[i]
        index=i

print(f"{max} is greatest element with {index} index")
```

```python
# second largest number

list1 = [12, 45, 64, 12, 556, 78, 65, 32]

largest = list1[0]
second_largest = list1[1]

for i in range(len(list1)):
    if list1[i] > largest:
        second_largest = largest
        largest = list1[i]
    elif list1[i] > second_largest and list1[i] != largest:
        second_largest = list1[i]

print("Largest:", largest)
print("Second largest:", second_largest)
```
